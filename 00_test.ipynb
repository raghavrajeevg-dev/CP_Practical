```python
def get_grade(marks):
    """Get grade based on marks"""
    if marks > 95:
        return 'A'
    elif marks > 85:
        return 'B'
    elif marks > 75:
        return 'C'
    elif marks > 65:
        return 'D'
    elif marks > 55:
        return 'E'
    else:
        return 'F'

# Get marks from student and display grade
marks = int(input("Enter your marks: "))
print(f"Grade {get_grade(marks)}")   
```
